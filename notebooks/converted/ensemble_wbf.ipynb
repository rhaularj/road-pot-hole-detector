````python
%%writefile utils.py
"""Shared helpers for the road damage detection scripts.

Not meant to be run directly. The other scripts in this folder import from it.
"""
from __future__ import annotations

import json
import random
from pathlib import Path

import numpy as np
import yaml

PROJECT_ROOT = Path(__file__).resolve().parents[1]
CLASS_NAMES = {0: "D00", 1: "D10", 2: "D20", 3: "D40"}
IMG_EXTS = {".jpg", ".jpeg", ".png", ".bmp"}
DEFAULT_THRESHOLDS = np.round(np.arange(0.05, 0.96, 0.05), 2)


# --------------------------------------------------------------------------- #
# General helpers
# --------------------------------------------------------------------------- #
def set_seed(seed: int) -> None:
    random.seed(seed)
    np.random.seed(seed)
    try:
        import torch

        torch.manual_seed(seed)
        torch.cuda.manual_seed_all(seed)
    except ImportError:
        pass


def save_json(obj, path) -> None:
    path = Path(path)
    path.parent.mkdir(parents=True, exist_ok=True)
    with open(path, "w", encoding="utf-8") as f:
        json.dump(obj, f, indent=2, default=float)


def resolve_weights(weights) -> str:
    """Use the path as given, or relative to the project root. Otherwise return it
    unchanged (so names like 'yolov8s.pt' can be auto-downloaded by Ultralytics)."""
    p = Path(weights)
    if p.exists():
        return str(p)
    if (PROJECT_ROOT / p).exists():
        return str(PROJECT_ROOT / p)
    return str(weights)


def _is_rtdetr(weights: str) -> bool:
    if "rtdetr" in weights.lower() or "rt-detr" in weights.lower():
        return True
    p = Path(weights)
    if p.suffix == ".pt" and p.exists():
        try:
            import torch

            ckpt = torch.load(str(p), map_location="cpu", weights_only=False)
            m = ckpt.get("ema") or ckpt.get("model")
            return type(m).__name__ == "RTDETRDetectionModel"
        except Exception:
            return False
    return False


def load_model(weights, task: str | None = "detect"):
    """Load a YOLO or RT-DETR model. RT-DETR needs its own class because it uses a
    different predictor, so it is detected from the file name or the checkpoint."""
    from ultralytics import RTDETR, YOLO

    weights = resolve_weights(weights)
    if _is_rtdetr(weights):
        return RTDETR(weights)
    return YOLO(weights, task=task) if task else YOLO(weights)


# --------------------------------------------------------------------------- #
# Dataset yaml helpers
# --------------------------------------------------------------------------- #
def resolve_data_yaml(data) -> Path:
    """Write a copy of the dataset yaml with an absolute `path:`.

    Ultralytics resolves relative dataset paths against its own datasets directory,
    not the current folder, so we always hand it an absolute path.
    """
    data_path = Path(data)
    if not data_path.exists():
        data_path = PROJECT_ROOT / data
    if not data_path.exists():
        raise FileNotFoundError(f"Dataset yaml not found: {data}")
    with open(data_path, encoding="utf-8") as f:
        cfg = yaml.safe_load(f)
    root = Path(cfg.get("path", data_path.parent))
    if not root.is_absolute():
        root = (PROJECT_ROOT / root).resolve()
    cfg["path"] = root.as_posix()
    out_dir = PROJECT_ROOT / "runs" / "_resolved"
    out_dir.mkdir(parents=True, exist_ok=True)
    out = out_dir / data_path.name
    with open(out, "w", encoding="utf-8") as f:
        yaml.safe_dump(cfg, f, sort_keys=False)
    return out


def load_names(data_yaml) -> dict[int, str]:
    with open(data_yaml, encoding="utf-8") as f:
        cfg = yaml.safe_load(f)
    names = cfg["names"]
    if isinstance(names, list):
        return {i: n for i, n in enumerate(names)}
    return {int(k): v for k, v in names.items()}


def get_split_dirs(data_yaml, split: str) -> tuple[Path, Path]:
    """Return (images_dir, labels_dir) for a split of a resolved dataset yaml."""
    with open(data_yaml, encoding="utf-8") as f:
        cfg = yaml.safe_load(f)
    if split not in cfg:
        raise KeyError(f"Split '{split}' is not defined in {data_yaml}")
    rel = cfg[split]
    if isinstance(rel, list):
        rel = rel[0]
    img_dir = Path(cfg["path"]) / rel
    parts = list(img_dir.parts)
    for i in range(len(parts) - 1, -1, -1):
        if parts[i] == "images":
            parts[i] = "labels"
            break
    else:
        raise ValueError(f"Cannot derive the labels folder from {img_dir}")
    return img_dir, Path(*parts)


def list_images(folder) -> list[Path]:
    return sorted(p for p in Path(folder).iterdir() if p.suffix.lower() in IMG_EXTS)


def read_yolo_labels(path) -> tuple[np.ndarray, np.ndarray]:
    """Read a YOLO label file. Returns (class_ids[n], boxes_xyxy_normalised[n, 4])."""
    path = Path(path)
    if not path.exists() or path.stat().st_size == 0:
        return np.zeros(0, dtype=int), np.zeros((0, 4))
    arr = np.loadtxt(path, ndmin=2)
    cls = arr[:, 0].astype(int)
    cx, cy, w, h = arr[:, 1], arr[:, 2], arr[:, 3], arr[:, 4]
    xyxy = np.stack([cx - w / 2, cy - h / 2, cx + w / 2, cy + h / 2], axis=1)
    return cls, xyxy


# --------------------------------------------------------------------------- #
# Detection metrics (F1 at IoU 0.5, the RDD challenge metric)
# --------------------------------------------------------------------------- #
def box_iou_np(a: np.ndarray, b: np.ndarray) -> np.ndarray:
    """Pairwise IoU between boxes a[n, 4] and b[m, 4] in xyxy format."""
    area_a = (a[:, 2] - a[:, 0]) * (a[:, 3] - a[:, 1])
    area_b = (b[:, 2] - b[:, 0]) * (b[:, 3] - b[:, 1])
    lt = np.maximum(a[:, None, :2], b[None, :, :2])
    rb = np.minimum(a[:, None, 2:], b[None, :, 2:])
    wh = np.clip(rb - lt, 0, None)
    inter = wh[..., 0] * wh[..., 1]
    union = area_a[:, None] + area_b[None, :] - inter
    return inter / np.maximum(union, 1e-12)


class DetectionEvaluator:
    """Accumulates predictions over a dataset and computes P / R / F1 at any
    confidence threshold.

    Predictions are matched to ground truth once (greedy, by descending confidence,
    same class, IoU >= iou_thr). Because greedy matching only depends on higher
    scoring boxes, the result for any threshold is just a filter on that single
    matching, so sweeping thresholds is cheap.

    IoU is computed on normalised coordinates, which gives the same value as pixel
    coordinates (IoU is invariant to scaling each axis).
    """

    def __init__(self, nc: int, iou_thr: float = 0.5):
        self.nc = nc
        self.iou_thr = iou_thr
        self.n_gt = np.zeros(nc, dtype=int)
        self._scores, self._cls, self._tp = [], [], []
        self._cache = None

    def add(self, pred_boxes, pred_scores, pred_cls, gt_boxes, gt_cls) -> None:
        self._cache = None
        pred_boxes = np.asarray(pred_boxes, dtype=float).reshape(-1, 4)
        pred_scores = np.asarray(pred_scores, dtype=float).reshape(-1)
        pred_cls = np.asarray(pred_cls).astype(int).reshape(-1)
        gt_boxes = np.asarray(gt_boxes, dtype=float).reshape(-1, 4)
        gt_cls = np.asarray(gt_cls).astype(int).reshape(-1)

        for c in range(self.nc):
            g = gt_boxes[gt_cls == c]
            self.n_gt[c] += len(g)
            m = pred_cls == c
            if not m.any():
                continue
            b, s = pred_boxes[m], pred_scores[m]
            order = np.argsort(-s)
            b, s = b[order], s[order]
            tp = np.zeros(len(b), dtype=bool)
            if len(g):
                ious = box_iou_np(b, g)
                matched = np.zeros(len(g), dtype=bool)
                for i in range(len(b)):
                    cand = np.where(matched, -1.0, ious[i])
                    j = int(cand.argmax())
                    if cand[j] >= self.iou_thr:
                        matched[j] = True
                        tp[i] = True
                        if matched.all():
                            break
            self._scores.append(s)
            self._cls.append(np.full(len(b), c, dtype=int))
            self._tp.append(tp)

    def _arrays(self):
        if self._cache is None:
            if self._scores:
                self._cache = (
                    np.concatenate(self._scores),
                    np.concatenate(self._cls),
                    np.concatenate(self._tp),
                )
            else:
                self._cache = (np.zeros(0), np.zeros(0, dtype=int), np.zeros(0, dtype=bool))
        return self._cache

    @staticmethod
    def _prf(tp: int, fp: int, fn: int) -> tuple[float, float, float]:
        p = tp / (tp + fp) if (tp + fp) else 0.0
        r = tp / (tp + fn) if (tp + fn) else 0.0
        f1 = 2 * p * r / (p + r) if (p + r) else 0.0
        return p, r, f1

    def at_threshold(self, conf: float) -> dict:
        scores, cls, tp = self._arrays()
        keep = scores >= conf
        tp_c = np.bincount(cls[keep & tp], minlength=self.nc)
        fp_c = np.bincount(cls[keep & ~tp], minlength=self.nc)
        fn_c = self.n_gt - tp_c
        per_class = []
        for c in range(self.nc):
            p, r, f1 = self._prf(int(tp_c[c]), int(fp_c[c]), int(fn_c[c]))
            per_class.append(
                {"class_id": c, "precision": p, "recall": r, "f1": f1,
                 "tp": int(tp_c[c]), "fp": int(fp_c[c]), "fn": int(fn_c[c]),
                 "n_gt": int(self.n_gt[c])}
            )
        p, r, f1 = self._prf(int(tp_c.sum()), int(fp_c.sum()), int(fn_c.sum()))
        return {"conf": float(conf), "precision": p, "recall": r, "f1": f1,
                "tp": int(tp_c.sum()), "fp": int(fp_c.sum()), "fn": int(fn_c.sum()),
                "per_class": per_class}

    def average_precision(self) -> np.ndarray:
        """Per-class AP at the evaluator's IoU threshold (101-point interpolation, as in COCO)."""
        scores, cls, tp = self._arrays()
        ap = np.zeros(self.nc)
        recall_points = np.linspace(0, 1, 101)
        for c in range(self.nc):
            if self.n_gt[c] == 0:
                continue
            m = cls == c
            order = np.argsort(-scores[m], kind="stable")
            hits = tp[m][order]
            tpc = np.cumsum(hits)
            fpc = np.cumsum(~hits)
            recall = tpc / self.n_gt[c]
            precision = tpc / np.maximum(tpc + fpc, 1)
            # precision envelope, then sample it at fixed recall points
            envelope = np.maximum.accumulate(precision[::-1])[::-1] if len(precision) else precision
            idx = np.searchsorted(recall, recall_points, side="left")
            ap[c] = np.mean([envelope[i] if i < len(envelope) else 0.0 for i in idx])
        return ap

    def sweep(self, thresholds=DEFAULT_THRESHOLDS) -> list[dict]:
        return [self.at_threshold(t) for t in thresholds]

    def best(self, thresholds=DEFAULT_THRESHOLDS):
        """Returns (best_overall, best_per_class, all_rows)."""
        rows = self.sweep(thresholds)
        best = max(rows, key=lambda r: r["f1"])
        best_pc = []
        for c in range(self.nc):
            br = max(rows, key=lambda r: r["per_class"][c]["f1"])
            best_pc.append({"conf": br["conf"], **br["per_class"][c]})
        return best, best_pc, rows


# --------------------------------------------------------------------------- #
# Results logging
# --------------------------------------------------------------------------- #
def append_metrics_row(row: dict, path=None) -> Path:
    """Append one row to results/metrics.csv (created if missing)."""
    import pandas as pd

    path = Path(path) if path else PROJECT_ROOT / "results" / "metrics.csv"
    path.parent.mkdir(parents=True, exist_ok=True)
    new = pd.DataFrame([row])
    if path.exists() and path.stat().st_size > 0:
        new = pd.concat([pd.read_csv(path), new], ignore_index=True)
    new.to_csv(path, index=False)
    return path
````

# `ensemble_wbf.py`

```text
Ensemble several detectors with Weighted Boxes Fusion (WBF), optionally with TTA.

This is the trick most top CRDDC2022 teams relied on. Each model predicts on the
split, the predictions are fused with WBF, and F1 @ IoU 0.5 is computed for every
single model and for the ensemble so you can see the gain directly.

Examples:
    python scripts/ensemble_wbf.py \
        --weights runs/detect/yolov8m_1024_s42/weights/best.pt runs/detect/yolo11m_1024_s42/weights/best.pt \
        --imgsz 1024 1024 --split val

    # with Ultralytics TTA (flip + multi-scale) and a sweep of the WBF IoU threshold
    python scripts/ensemble_wbf.py --weights a.pt b.pt c.pt --tta --wbf-iou 0.45 0.55 0.65

    # tune on val, then report on test with the threshold chosen on val
    python scripts/ensemble_wbf.py --weights a.pt b.pt --split test --fixed-conf 0.30

Notes:
  * Models run at low confidence (0.001) so WBF has all candidates; the confidence
    threshold is chosen afterwards by the F1 sweep.
  * --tta has no effect on RT-DETR (Ultralytics falls back to single-scale).
  * Selecting the WBF IoU / threshold on val and reporting on a separate test split
    avoids over-tuning on one set.
```

````python
import argparse
from pathlib import Path

import numpy as np
from tqdm import tqdm

from utils import (DEFAULT_THRESHOLDS, PROJECT_ROOT, DetectionEvaluator, append_metrics_row, get_split_dirs,
                   list_images, load_model, load_names, read_yolo_labels, resolve_data_yaml, resolve_weights,
                   save_json)
````

````python
try:
    from ensemble_boxes import weighted_boxes_fusion
except ImportError as e:  # pragma: no cover
    raise SystemExit("ensemble-boxes is not installed. Run: pip install ensemble-boxes") from e
````

````python
def parse_args():
    p = argparse.ArgumentParser(description=__doc__, formatter_class=argparse.RawDescriptionHelpFormatter)
    p.add_argument("--weights", nargs="+", required=True, help="two or more trained models")
    p.add_argument("--data", default="configs/rdd2022.yaml")
    p.add_argument("--split", default="val")
    p.add_argument("--imgsz", type=int, nargs="+", default=[640], help="one value, or one per model")
    p.add_argument("--batch", type=int, default=16)
    p.add_argument("--device", default=None)
    p.add_argument("--conf", type=float, default=0.001)
    p.add_argument("--nms-iou", type=float, default=0.6)
    p.add_argument("--tta", action="store_true", help="Ultralytics test-time augmentation for each model")
    p.add_argument("--wbf-iou", type=float, nargs="+", default=[0.55], help="WBF IoU threshold(s) to try")
    p.add_argument("--skip-thr", type=float, default=0.001, help="WBF: ignore boxes below this score")
    p.add_argument("--conf-type", default="avg", choices=["avg", "max", "box_and_model_avg", "absent_model_aware_avg"])
    p.add_argument("--model-weights", type=float, nargs="+", default=None, help="relative weight per model")
    p.add_argument("--fixed-conf", type=float, default=None, help="also report F1 at this exact threshold")
    p.add_argument("--name", default="ensemble")
    p.add_argument("--max-images", type=int, default=None)
    return p.parse_args()
````

````python
def predict_all(weights: str, imgsz: int, images, args):
    """Run one model over all images. Returns {image_path: (boxes_xyxyn, scores, labels)}."""
    model = load_model(weights)
    out = {}
    stream = model.predict(source=[str(p) for p in images], imgsz=imgsz, conf=args.conf, iou=args.nms_iou,
                           device=args.device, batch=args.batch, augment=args.tta, stream=True, verbose=False)
    for r in tqdm(stream, total=len(images), desc=f"Predicting ({Path(weights).parent.parent.name or weights})"):
        b = r.boxes
        out[str(r.path)] = (b.xyxyn.cpu().numpy(), b.conf.cpu().numpy(), b.cls.cpu().numpy().astype(int))
    return out
````

````python
def fuse(per_model, key, args, iou_thr):
    boxes, scores, labels = [], [], []
    for pm in per_model:
        b, s, l = pm[key]
        boxes.append(np.clip(b, 0, 1).tolist())
        scores.append(s.tolist())
        labels.append(l.tolist())
    if all(len(s) == 0 for s in scores):
        return np.zeros((0, 4)), np.zeros(0), np.zeros(0, dtype=int)
    fb, fs, fl = weighted_boxes_fusion(boxes, scores, labels, weights=args.model_weights, iou_thr=iou_thr,
                                       skip_box_thr=args.skip_thr, conf_type=args.conf_type)
    return fb, fs, np.asarray(fl).astype(int)
````

````python
def evaluate(pred_fn, images, lbl_dir, nc):
    ev = DetectionEvaluator(nc, iou_thr=0.5)
    for p in images:
        gt_cls, gt_boxes = read_yolo_labels(lbl_dir / f"{p.stem}.txt")
        b, s, l = pred_fn(str(p))
        ev.add(b, s, l, gt_boxes, gt_cls)
    return ev
````

````python
def main():
    args = parse_args()
    n = len(args.weights)
    if n < 2:
        raise SystemExit("Give at least two models to --weights.")
    imgsz = args.imgsz * n if len(args.imgsz) == 1 else args.imgsz
    if len(imgsz) != n:
        raise SystemExit("--imgsz must be one value or one value per model.")
    if args.model_weights and len(args.model_weights) != n:
        raise SystemExit("--model-weights must have one value per model.")

    data_yaml = resolve_data_yaml(args.data)
    nc = len(load_names(data_yaml))
    names = load_names(data_yaml)
    img_dir, lbl_dir = get_split_dirs(data_yaml, args.split)
    images = list_images(img_dir)
    if args.max_images:
        images = images[:args.max_images]
    if not images:
        raise SystemExit(f"No images found in {img_dir}")

    weights = [resolve_weights(w) for w in args.weights]
    per_model = [predict_all(w, s, images, args) for w, s in zip(weights, imgsz)]
    labels = [Path(w).parent.parent.name or Path(w).stem for w in weights]

    results = []  # (label, evaluator)
    for lab, pm in zip(labels, per_model):
        results.append((lab, evaluate(lambda k, pm=pm: pm[k], images, lbl_dir, nc)))
    for thr in args.wbf_iou:
        ev = evaluate(lambda k, thr=thr: fuse(per_model, k, args, thr), images, lbl_dir, nc)
        results.append((f"{args.name}_wbf{thr:g}" + ("_tta" if args.tta else ""), ev))

    # ---------- report ----------
    print("\n" + "=" * 78)
    print(f"Split: {args.split} | images: {len(images)} | TTA: {args.tta}")
    print("=" * 78)
    print(f"{'model':<44}{'best F1':>9}{'conf':>7}{'P':>7}{'R':>7}" + (f"{'F1@fixed':>10}" if args.fixed_conf else ""))
    summary = {}
    for lab, ev in results:
        best, best_pc, _ = ev.best(DEFAULT_THRESHOLDS)
        fixed = ev.at_threshold(args.fixed_conf) if args.fixed_conf is not None else None
        line = f"{lab:<44}{best['f1']:>9.4f}{best['conf']:>7.2f}{best['precision']:>7.3f}{best['recall']:>7.3f}"
        if fixed:
            line += f"{fixed['f1']:>10.4f}"
        print(line)
        summary[lab] = {"best": {k: v for k, v in best.items() if k != "per_class"},
                        "per_class_f1": {names[c]: best_pc[c]["f1"] for c in range(nc)},
                        "fixed": ({k: v for k, v in fixed.items() if k != "per_class"} if fixed else None)}
        if lab.startswith(args.name):
            row = {"run": lab, "split": args.split, "imgsz": "/".join(map(str, imgsz)), "precision": best["precision"],
                   "recall": best["recall"], "f1": best["f1"], "f1_conf": best["conf"]}
            if fixed:
                row.update({"fixed_conf": fixed["conf"], "f1_at_fixed_conf": fixed["f1"]})
            for c in range(nc):
                row[f"f1_{names[c]}"] = best_pc[c]["f1"]
            append_metrics_row(row)

    out_path = PROJECT_ROOT / "results" / "evaluations" / f"{args.name}_{args.split}_ensemble.json"
    save_json({"weights": weights, "imgsz": imgsz, "tta": args.tta, "wbf_iou": args.wbf_iou,
               "model_weights": args.model_weights, "conf_type": args.conf_type, "results": summary}, out_path)
    print(f"\nSaved {out_path}\nEnsemble rows appended to results/metrics.csv")
````
